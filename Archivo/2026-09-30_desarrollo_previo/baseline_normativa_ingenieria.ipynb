Evaluación de **Qwen3-4B** mediante *prompting* directo, sin recuperación de documentos (RAG), sobre un conjunto de 50 preguntas con respuesta de referencia verificada contra la normativa oficial. Este baseline cuantifica el desempeño del modelo previo a la aplicación de cualquier técnica.

## 3. Carga del modelo (Qwen3-4B, Apache 2.0)

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

MODEL_NAME = "Qwen/Qwen3-4B"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME, torch_dtype="auto", device_map="auto")
print(MODEL_NAME, "|", round(torch.cuda.memory_allocated()/1e9, 2), "GB VRAM")
```

## 4. Configuración de inferencia

*Prompting* directo, sin entregar los documentos del corpus, con decodificación determinista. El *system prompt* solicita el dato exacto y la cita del artículo, e instruye abstenerse cuando la información no está disponible.

```python
SYSTEM = (
    "Eres un asistente experto en la normativa de pregrado de la Facultad de "
    "Ingeniería de la Universidad de Concepción. Responde de forma breve, con el "
    "dato exacto y citando el artículo correspondiente (por ejemplo: 'Art. 8'). "
    "Si no tienes la información, di explícitamente que no está en la normativa."
)

def preguntar(pregunta, sistema=SYSTEM, max_new_tokens=256):
    messages = [{"role": "system", "content": sistema},
                {"role": "user", "content": pregunta}]
    text = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True, enable_thinking=False)
    inputs = tokenizer([text], return_tensors="pt").to(model.device)
    with torch.no_grad():
        out = model.generate(**inputs, max_new_tokens=max_new_tokens, do_sample=False)
    return tokenizer.decode(out[0][inputs.input_ids.shape[-1]:], skip_special_tokens=True).strip()
```

## 6. Ejecución del baseline

```python
import pandas as pd
pd.set_option("display.max_colwidth", None)  # mostrar la respuesta completa, sin recortar

registros = []
for p in preguntas:
    r = preguntar(p["pregunta"])
    registros.append({"categoria": p["categoria"], "pregunta": p["pregunta"],
                      "respuesta_modelo": r, "gold_dato": p["gold_dato"],
                      "gold_fuente": p["gold_fuente"]})

df = pd.DataFrame(registros)
df.to_csv("respuestas_baseline.csv", index=False)
df[["categoria", "respuesta_modelo", "gold_dato", "gold_fuente"]]
```

## 7. Calificación y resultados

Calificación estricta: una respuesta es correcta solo si el dato **y** el artículo son correctos; en las preguntas de premisa falsa, es correcta si el modelo reconoce que el dato no existe. Resultados agregados por categoría.

```python
veredictos = ['no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'no', 'si', 'si']

df["correcto"] = veredictos
df.to_csv("resultados_baseline.csv", index=False)

df["acierto"] = df["correcto"].str.strip().str.lower().isin(["si", "sí"])
resumen = df.groupby("categoria")["acierto"].agg(["sum", "count"])
resumen["exactitud_%"] = (100*resumen["sum"]/resumen["count"]).round(0)
orden = ["factual", "numerica", "condicional", "cruce", "abstencion"]
print(resumen.reindex(orden))
print("\nExactitud global: {:.0f}%".format(100*df["acierto"].mean()))

con_respuesta = df[df["categoria"] != "abstencion"]
print("Aciertos en preguntas con respuesta en el corpus: {}/{}".format(
    int(con_respuesta["acierto"].sum()), len(con_respuesta)))
```
